# round / floor /ceil:

# 1. round:

```python
# The numpy.round() function rounds the elements of an array to the nearest integer 
# or to the specified number of decimals.
```

```python
# Round to the nearest integer
arr = np.array([1.2, 2.7, 3.5, 4.9]) 
rounded_arr = np.round(arr)
print(rounded_arr)
```

```python
# Round to two decimals
arr = np.array([1.234, 2.567, 3.891]) 
rounded_arr = np.round(arr, decimals=2) 
print(rounded_arr)
```

```python
#randomly
np.round(np.random.random((2,3))*100)
```

# 2. floor:

```python
# The numpy.floor() function returns the largest integer less than or equal to each-
# -element of an array.
```

```python
# Floor operation
arr = np.array([1.2, 2.7, 3.5, 4.9]) 
floored_arr = np.floor(arr)
print(floored_arr)
```

```python
np.floor(np.random.random((2,3))*100) # gives the smallest integer 
```

# 3. Ceil:

```python
# The numpy.ceil() function returns the smallest integer greater than or equal to each-
# -element of an array.
```

```python
arr = np.array([1.2, 2.7, 3.5, 4.9]) 
ceiled_arr = np.ceil(arr)
print(ceiled_arr)
```

```python
np.ceil(np.random.random((2,3))*100) # gives highest integer ex : 7.8 = 8
```

```python
w1
```

```python
w2
```
