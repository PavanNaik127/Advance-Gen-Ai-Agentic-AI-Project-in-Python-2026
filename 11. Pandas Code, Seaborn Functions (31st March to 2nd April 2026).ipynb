# **Seaborn:**

```python
import matplotlib.pyplot as plt
import seaborn as sns


%matplotlib inline
plt.rcParams['figure.figsize'] = 6,2

import warnings
warnings.filterwarnings('ignore') # os error
```

# distplot():

```python
vis1 = sns.distplot(df["InternetUsers"])
#plt.show()
# if we dont get Output, we can write plt.show() below the above code.
```

```python
# Univariant Analysis   -- Plot the graph using 1 variable
# BiVariant Analysis    -- Plot the graph using 2 variables
# MultiVariant Analysis -- Plot the graph using more than 2 variables
```

# displot():

```python
vis2 = sns.displot(df["InternetUsers"])
```

```python
# Displot(NEW)/Distplot(OLD) --> Distribution plot
# Displot: Only histogram → shows how many countries fall in each range
# Distplot: Histogram + KDE --> Show overall Distribution Shape
# displot → “How many values?” 📊
# distplot → “What is the shape?” 📈
# 👉 displot is a figure-level function
# 👉 distplot was an axes-level function
```

```python
vis3 = sns.distplot(df["InternetUsers"], bins = 15)
```

```python
vis3 = sns.displot(df["InternetUsers"], bins = 15)
```

# Box Plots:

```python
# Box Plots:  box-and-whisker plot:
vis4 = sns.boxplot(data= df, x = 'IncomeGroup', y = 'BirthRate') # Bi-Variant Analysys
# The Diagram Below has 3 Outliers. That Arrow indicates an Outlier. 
# An Outlier is a point that is very far from the remaining points.
# using graphs, we will find Outliers
```

```python
# OUTLIER: Statistics outlier is the datapoint that is far from other observation is called Outlier
# Outlier is also called as Anomaly Detection
# These outlier impact machine learning model accuracy
# To adjust these outlier we will introduce Probablity Concept
# which algorithm probability explains - 
```

# lmplot:

```python
vis5 = sns.lmplot(data = df, x = 'InternetUsers', y ='BirthRate') 
# lm - Linear Model Plot
```

# fit_reg:

```python
vis6 = sns.lmplot(data = df, x = 'InternetUsers', y ='BirthRate', fit_reg = False) 
# ""fit_reg = False"" will make the line disappear
```

# hue:

```python
vis7 = sns.lmplot(data = df, x = 'InternetUsers', y ='BirthRate', fit_reg = False, hue = 'IncomeGroup') 
# Hue - Parameter for color
```

```python
vis7 = sns.lmplot(data = df, x = 'InternetUsers', y ='BirthRate', fit_reg = True, hue = 'IncomeGroup') 
```
