# 1. While loop:

```python
i = 1          # initializing

while i<=5:    # condition
    print('data science')
    i = i + 1  # increment
```

```python
i = 1          # initializing

while i<=5:    # condition
    print('data science :', i)
    i = i + 1  # increment
```

```python
i = 5          # initializing

while i>=1:    # condition
    print('data science :', i)
    i = i - 1  # increment
```

```python
i = 1

while i<=5:
    print('data science') # when we mention end then new line will not create
    j = 1
    while j<=4:
        print('technology')
        j = j + 1
        
    i = i + 1
    print()    
# 1st i =1, then it will print data science. Then it will go to j. j<=4 satisfies, then it will be j =j+1(Total 4times).
# then it will go to i=i+1--> i = 2. This process continues till i=5.
```

# end = "":

```python
i = 1
while i<=5:
    print(' datascience', end = "") # when we mention end then new line will not create
    j = 1
    while j<=4:
        print(' technology', end= "--")
        j = j + 1
        
    i = i + 1
    print()
```

```python
i = 1

while i <= 2 :
    j = 0
    
    while  j <= 2 :
        print(i*j, end=" ")
        j += 1
    print()
    i += 1
```

```python
i = 1
while i <= 4 :
    j = 0
    while  j <= 3 :
        print(i*j, end="____")
        j += 1
    print()
    i += 1
```

# **3. Transfer Statements:**

```python
# LETS DISCUSS ABOUT 3 KEYWORDS -- BREAK || CONTINUE || PASS

# BREAK STATEMENT - if you apply break statement in a loop then it will end the loop
# Break = jumps out of the function/loop
# Pass = skips block of code (function, class etc)
# Continue = skips 1 step/iteration during loop
```

# Break:

```python
for i in range(1,11):
    print(i)
```

```python
for i in range(1,11):
    if i == 5:
       break           # It will stop at 5.
    print(i)
```

# Continue:

```python
# In continue, the loop won't be terminated, and excludes the assigned number, and gives all other values.
for i in range(1,11):
    if i == 8:
       continue      # It gives every number except 8.
    print(i)
```

```python
for i in range(1,11):
    if i == 6:
        break  
    print('hello :',i)    
```

# Pass:

```python
# for i in range(1,11): 
```

```python
for i in range(1,11): 
    pass   
# in the above 41st cell, we didn't give a pass, so it showed an error. But now we have given pass. 
# pass will skip/remove any error like the above one.
```

```python
for i in range(1,51):    
    if i%3 == 0:
    
        print(i)
print('end')    
```

```python
for i in range(1,51):    
    if i%3 != 0:  # not equal to 0
    
        print(i)
print('end')    
```

```python
for i in range(1,51):
    
    if i%3 == 0:
        continue
    print(i)
print('end')    
```

```python
for i in range(1,51):
    
    if i%3 == 0 or i%5 == 0:
        continue           # 3 and 5 multiples are skipped.
    print(i)
```
